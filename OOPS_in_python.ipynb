```python
class Result:
  def __init__(self,name,marks):
    self.name = name
    self.marks = marks

  @staticmethod
  def express_love():
    return print("Hello, Dear Student,")
  def get_average(self):
    sum = 0
    for i in self.marks:
      sum = sum + i
    return print(f" your average score is {sum/len(self.marks)}")

stu1 = Result("Shreya", [56,60,45])
stu1.express_love()
print(stu1.name)
print(stu1.marks)
stu1.get_average()
```

```python
class Account:
  def __init__(self,balance,account_no, amount):
    self.balance = balance
    self.account_no = account_no
    self.amount = amount

  def debit(self):
    self.balance = self.balance - self.amount
    return self.balance

  def credit(self):
    self.balance = self.balance + self.amount
    return self.balance

balance = int(input("Enter your balance: "))
account_no = int(input("Enter your account number: "))
amount = int(input("Enter the amount for debit or credit: "))
usr1 = Account(balance, account_no,amount)
print("Enter 0 for debit and enter 1 for credit to your balance.")
choice = int(input())

if choice == 0:
  print(usr1.debit())
elif choice == 1:
   print(usr1.credit())
else:
  print("Invalid choice.")
```

```python
#staticmethod is used when we use function without engraining  the self
#classmethod is used by using cls as a symbol for ingraining class within a function

#property  to use when we cannot always change have unaltered data or changing data.





# dunder functions for operator overloading

class Complex:
  def __init__(self, real,img):
    self.real = real
    self.img = img

  def show_complex(self):
    print(f"{self.real}i + {self.img}j")

  def __add__(self, num2):
    new_real = self.real + num2.real
    new_img = self.img + num2.img
    return Complex(new_real, new_img)


num1 = Complex(4,5)
num2 = Complex(8,9)
num1.show_complex()
num3 = num1 + num2
num3.show_complex()
```

```python
# ##Practicee problem 1
# to find the area of circle and perimeter using the oops in python

class Circle:
  def __init__(self, radius):
    self.radius = radius

  def area(self):
    return 3.14*(self.radius)**2

  def peri(self):
    return 2*3.14*self.radius

r = int(input("Enter the radius of the circle: "))
c1 = Circle(r)
print(f"The area of the circle is {c1.area()}")
print(f"The perimeter of the circle is {c1.peri()}")
```

```python
#practice problem to 2
# to create an Employee class with attributes like role , department salary and has show_details_method
#then create an Engineer class that inherites from the enployee class and has attributes of age and name


class Employee:
  def __init__(self, role, department, salary):
    self.role = role
    self.department = department
    self.salary = salary

  def show_details(self):
    print(f"The role of the employee is {self.role}")
    print(f"The department of the employee is {self.department}")
    print(f"The salary of the employee is {self.salary}")

s1 = Employee("Manager", "HR", 50000)
s1.show_details()
class Engineer(Employee):
  def __init__(self,name, age):
    self.name = name
    self.age = age
    super().__init__("CEO", "Tech and Operations", "1000k$")
    print(f"The name of the engineer is {self.name}")
    print(f"The age of the engineer is {self.age}")

s1 = Engineer("Mark", 17)
s1.show_details()

```

```python
#PROBLEM 3 of storing items and prices in a classes and objects using dunder functions


class Items:
  def __init__(self, item, prices):
    self.item = item
    self.prices = prices

  def __gt__(self, itm2):
    if self.prices > itm2.prices:
      return print(f"{self.item} is more expensive than {itm2.item}")
    else:
      return print(f"{itm2.item} is more expensive than {self.item}")


itm1 = Items("Creatine Monohydrate", 1000)
itm2 = Items("Performance Protein", 2500)
itm3 = itm1 >itm2
```
