```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LinearRegression


x = np.array([2025,2026,2027,2028])
y = np.array([15,1000,65,87])

plt.plot(x,y, marker = "X",markersize = 30, markerfacecolor = "cyan",markeredgecolor = "blue", color = "red", linestyle = "solid")
plt.xticks(x)
plt.show()
```

```python
X = np.array([2025,2026,2027,2028])
Y = np.array([50,60,70,80])

plt.plot(X,Y,marker =".",markersize = 40, markerfacecolor = "Orange", markeredgecolor = "green", color = "Black", linestyle = "dashed")
plt.xlabel("Year", color = "Green", fontsize = 20, fontweight = "bold")
plt.ylabel("Population", color = "Green", fontsize = 20, fontweight = "bold")
plt.title("Population Growth", color = "Green", fontsize = 20, fontweight = "bold")
plt.xticks(X)
plt.yticks(Y)
plt.tick_params(axis = "both", colors = "blue")
plt.show()
```

```python
x_1 = [1,2,3,4,5]
y_1 = [26,27,28,29,30]

plt.grid(axis = "both", linewidth = 3, color = "orange",linestyle = "dotted")
plt.plot(x_1,y_1,color = "Black")
plt.show()
```

```python
categories = np.array(["Fruits", "Vegetables", "Proteins", "Carbs"])
values = np.array([50,100,150,200,])

plt.bar(categories,values)
plt.title("Daily Consumption", fontsize = 14, fontweight = "bold")
plt.xlabel("Categories")
plt.ylabel("Consumption")
plt.tick_params(axis = "both", colors = "red")
plt.show()
#This is  a Bar Chart
```

```python
#This is a PIE CHART

categories = ["Juniors", "Freshmen", "Sophomores", "Seniors"]
amount = np.array([200,300,350,600])

plt.pie(amount, labels = categories, autopct = "%1.2f%%", explode = [0,0,0,0.2], shadow = True,startangle = 190)
plt.title("Student Enrollment 2026")

plt.show()
#learnt about pie chart
#plt.pie()function
#it includes three four parameters that are labels , values that we put in , explode to separate out a piece of data , "autopct" for percentage
```
